```python
# ==============================================================================
# 1. SETUP — Mount Drive, Instalacija i Učitavanje Skupa Podataka
# ==============================================================================
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
from torch.utils.data import Dataset, DataLoader, random_split
from torchvision import transforms
from transformers import AutoModel

from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, f1_score, mean_absolute_error, ConfusionMatrixDisplay
from statsmodels.stats.contingency_tables import mcnemar

from google.colab import drive
from huggingface_hub import login
from google.colab import userdata

# 1.1 Mount Google Drive
if not os.path.exists('/content/drive/MyDrive'):
    drive.mount('/content/drive')
else:
    print("Drive je već mount-ovan.")

# 1.2 Instalacija potrebnih paketa
!pip install -q datasets transformers statsmodels

# 1.3 HuggingFace Login (opciono)
try:
    hf_token = userdata.get('HF_TOKEN')
    login(token=hf_token)
except Exception:
    print("HF_TOKEN nije podešen — nastavljam bez eksplicitne prijave.")

# 1.4 Učitavanje FairFace skupa podataka
from datasets import load_dataset
ds = load_dataset("HuggingFaceM4/FairFace", "0.25")
print(ds)

# Direktorijumi za čuvanje rezultata
os.makedirs("/content/drive/MyDrive/diplomski2/features", exist_ok=True)
os.makedirs("/content/drive/MyDrive/diplomski2/results", exist_ok=True)

# Uređaj
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Koristi se uređaj:", device)


# ==============================================================================
# 2. DATASET & DATALOADERS — Definisanje i Podela Podataka (Train / Val / Test)
# ==============================================================================
class FairFaceDataset(Dataset):
    def __init__(self, hf_split, transform=None):
        self.data = hf_split
        self.transform = transform

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        primer = self.data[idx]
        slika = primer["image"].convert("RGB")
        if self.transform:
            slika = self.transform(slika)
        age = primer["age"]
        gender = primer["gender"]
        race = primer["race"]
        return slika, age, gender, race

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

train_dataset = FairFaceDataset(ds["train"], transform=transform)
val_full = FairFaceDataset(ds["validation"], transform=transform)

# Podela validacionog skupa na pola (pola za Val tuning, pola za Test)
val_size = len(val_full) // 2
test_size = len(val_full) - val_size
val_dataset, test_dataset = random_split(val_full, [val_size, test_size])

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True, num_workers=2, pin_memory=True)
val_loader = DataLoader(val_dataset, batch_size=64, shuffle=False, num_workers=2, pin_memory=True)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False, num_workers=2, pin_memory=True)

print(f"Train: {len(train_dataset)} | Val: {len(val_dataset)} | Test: {len(test_dataset)}")


# ==============================================================================
# 3. EKSTRAKCIJA FEATURE-A — ConvNeXt, DeiT Distilled i Swin
# ==============================================================================
def ucitaj_model(model_name, device):
    model = AutoModel.from_pretrained(model_name).to(device)
    model.eval()
    for p in model.parameters():
        p.requires_grad = False
    return model

def izvuci_feature_i_labele(loader, model, device, ime_prefix="model_train", nacin_ekstrakcije="pooler", print_svakih=200):
    sve_feature, sve_age, sve_gender, sve_race = [], [], [], []
    ukupno_slika = len(loader.dataset)

    for i, (slike, ages, genders, races) in enumerate(loader):
        with torch.no_grad():
            izlaz = model(slike.to(device))

            if nacin_ekstrakcije == "deit_distilled":
                # Usrednjavanje [CLS] i [DISTILLATION] tokena
                cls_tok = izlaz.last_hidden_state[:, 0, :]
                dist_tok = izlaz.last_hidden_state[:, 1, :]
                feat = ((cls_tok + dist_tok) / 2.0).cpu().numpy()
            elif nacin_ekstrakcije == "cls":
                feat = izlaz.last_hidden_state[:, 0, :].cpu().numpy()
            else:
                # Pooler output za ConvNeXt i Swin
                feat = izlaz.pooler_output.cpu().numpy()

            sve_feature.append(feat)
            sve_age.append(ages.numpy())
            sve_gender.append(genders.numpy())
            sve_race.append(races.numpy())

        # Ispis progresa radi sigurnosti u Colab-u
        if (i + 1) % print_svakih == 0 or (i + 1) == len(loader):
            obradjeno = min((i + 1) * loader.batch_size, ukupno_slika)
            print(f"[{ime_prefix}] Obrađeno {obradjeno}/{ukupno_slika} slika ({(obradjeno/ukupno_slika)*100:.1f}%)")

    return (np.concatenate(sve_feature), np.concatenate(sve_age),
            np.concatenate(sve_gender), np.concatenate(sve_race))

def sacuvaj_feature(prefix, feat, age, gender, race):
    np.save(f"/content/drive/MyDrive/diplomski2/features/{prefix}_features.npy", feat)
    np.save(f"/content/drive/MyDrive/diplomski2/features/{prefix}_age.npy", age)
    np.save(f"/content/drive/MyDrive/diplomski2/features/{prefix}_gender.npy", gender)
    np.save(f"/content/drive/MyDrive/diplomski2/features/{prefix}_race.npy", race)

modeli_konfig = {
    "convnext": {"path": "facebook/convnext-base-224-22k", "nacin": "pooler"},
    "deit": {"path": "facebook/deit-base-distilled-patch16-224", "nacin": "deit_distilled"},
    "swin": {"path": "microsoft/swin-base-patch4-window7-224-in22k", "nacin": "pooler"},
}

for ime, cfg in modeli_konfig.items():
    print(f"\n--- Započinjem ekstrakciju za model: {ime.upper()} ---")
    model = ucitaj_model(cfg["path"], device)

    for split_ime, loader in [("train", train_loader), ("val", val_loader), ("test", test_loader)]:
        feat, age, gender, race = izvuci_feature_i_labele(
            loader, model, device, ime_prefix=f"{ime}_{split_ime}", nacin_ekstrakcije=cfg["nacin"]
        )
        sacuvaj_feature(f"{ime}_{split_ime}", feat, age, gender, race)
        print(f"Uspešno sačuvano: {ime}_{split_ime}")

    del model
    torch.cuda.empty_cache()

print("\nEkstrakcija svih feature-a je uspešno završena!")


# ==============================================================================
# 4. KLASIFIKACIJA I REGRESIJA — StandardScaler i Val Tuning
# ==============================================================================
age_middle = {0: 1, 1: 6, 2: 14.5, 3: 24.5, 4: 34.5, 5: 44.5, 6: 54.5, 7: 64.5, 8: 75}

def treniraj_i_evaluiraj(ime_modela):
    X_train = np.load(f"/content/drive/MyDrive/diplomski2/features/{ime_modela}_train_features.npy")
    y_train = np.load(f"/content/drive/MyDrive/diplomski2/features/{ime_modela}_train_age.npy")

    X_val = np.load(f"/content/drive/MyDrive/diplomski2/features/{ime_modela}_val_features.npy")
    y_val = np.load(f"/content/drive/MyDrive/diplomski2/features/{ime_modela}_val_age.npy")

    X_test = np.load(f"/content/drive/MyDrive/diplomski2/features/{ime_modela}_test_features.npy")
    y_test = np.load(f"/content/drive/MyDrive/diplomski2/features/{ime_modela}_test_age.npy")

    y_train_years = np.array([age_middle[a] for a in y_train])
    y_test_years = np.array([age_middle[a] for a in y_test])

    rezultati = {}
    predikcije = {}

    # 4.1 Logistic Regression (Tuning C na Val)
    best_c_logreg, best_val_acc = 1.0, 0.0
    for c in [0.01, 0.1, 1.0, 10.0]:
        pipe = Pipeline([('scaler', StandardScaler()), ('clf', LogisticRegression(C=c, max_iter=2000))])
        pipe.fit(X_train, y_train)
        acc = pipe.score(X_val, y_val)
        if acc > best_val_acc:
            best_val_acc = acc
            best_c_logreg = c

    final_logreg = Pipeline([('scaler', StandardScaler()), ('clf', LogisticRegression(C=best_c_logreg, max_iter=2000))])
    final_logreg.fit(X_train, y_train)
    pred_logreg = final_logreg.predict(X_test)

    rezultati["LogisticRegression"] = {
        "accuracy": accuracy_score(y_test, pred_logreg),
        "f1_macro": f1_score(y_test, pred_logreg, average="macro"),
        "mae": None,
    }
    predikcije["LogisticRegression"] = pred_logreg

    # 4.2 Linear SVM (Tuning C na Val)
    best_c_svm, best_val_acc_svm = 1.0, 0.0
    for c in [0.001, 0.01, 0.1, 1.0]:
        pipe = Pipeline([('scaler', StandardScaler()), ('clf', LinearSVC(C=c, max_iter=5000, dual=False))])
        pipe.fit(X_train, y_train)
        acc = pipe.score(X_val, y_val)
        if acc > best_val_acc_svm:
            best_val_acc_svm = acc
            best_c_svm = c

    final_svm = Pipeline([('scaler', StandardScaler()), ('clf', LinearSVC(C=best_c_svm, max_iter=5000, dual=False))])
    final_svm.fit(X_train, y_train)
    pred_svm = final_svm.predict(X_test)

    rezultati["SVM"] = {
        "accuracy": accuracy_score(y_test, pred_svm),
        "f1_macro": f1_score(y_test, pred_svm, average="macro"),
        "mae": None,
    }
    predikcije["SVM"] = pred_svm

    # 4.3 Linear Regression
    final_linreg = Pipeline([('scaler', StandardScaler()), ('reg', LinearRegression())])
    final_linreg.fit(X_train, y_train_years)
    pred_linreg = final_linreg.predict(X_test)

    rezultati["LinearRegression"] = {
        "accuracy": None,
        "f1_macro": None,
        "mae": mean_absolute_error(y_test_years, pred_linreg),
    }
    predikcije["LinearRegression"] = pred_linreg

    for klasifikator, p in predikcije.items():
        np.save(f"/content/drive/MyDrive/diplomski2/features/{ime_modela}_pred_{klasifikator}.npy", p)

    return rezultati

sve_metrike = []
for ime_modela in ["convnext", "deit", "swin"]:
    print(f"Obučavam klasifikatore za {ime_modela}...")
    rezultati = treniraj_i_evaluiraj(ime_modela)
    for klasifikator, metrike in rezultati.items():
        sve_metrike.append({"model": ime_modela, "klasifikator": klasifikator, **metrike})

tabela_rezultata = pd.DataFrame(sve_metrike)
tabela_rezultata.to_csv("/content/drive/MyDrive/diplomski2/results/sumarna_tabela.csv", index=False)
print("\n=== SUMARNA TABELA REZULTATA ===")
print(tabela_rezultata)


# ==============================================================================
# 5. FAIRNESS ANALIZA — Pojedinačna i Intersekcionalna (Rasa x Pol)
# ==============================================================================
race_names = ["East Asian", "Indian", "Black", "White", "Middle Eastern", "Latino_Hispanic", "Southeast Asian"]
gender_names = ["Male", "Female"]

race_test = np.load("/content/drive/MyDrive/diplomski2/features/convnext_test_race.npy")
gender_test = np.load("/content/drive/MyDrive/diplomski2/features/convnext_test_gender.npy")

intersekcija_test = np.array([f"{race_names[r]} - {gender_names[g]}" for r, g in zip(race_test, gender_test)])
intersekcija_imena = sorted(list(set(intersekcija_test)))

def metrike_po_grupi(y_true, y_pred, grupa_labele, grupa_imena, ime_dimenzije):
    rezultati = []
    for ime in grupa_imena:
        maska = grupa_labele == ime
        if maska.sum() == 0:
            continue
        acc = accuracy_score(y_true[maska], y_pred[maska])
        f1 = f1_score(y_true[maska], y_pred[maska], average="macro")
        rezultati.append({
            "dimenzija": ime_dimenzije,
            "grupa": ime,
            "n_uzoraka": int(maska.sum()),
            "accuracy": acc,
            "f1_macro": f1,
        })
    return pd.DataFrame(rezultati)

sve_fairness_metrike = []
for ime_modela in ["convnext", "deit", "swin"]:
    y_test = np.load(f"/content/drive/MyDrive/diplomski2/features/{ime_modela}_test_age.npy")
    for klasifikator in ["LogisticRegression", "SVM"]:
        pred = np.load(f"/content/drive/MyDrive/diplomski2/features/{ime_modela}_pred_{klasifikator}.npy")

        df_rasa = metrike_po_grupi(y_test, pred, np.array([race_names[r] for r in race_test]), race_names, "race")
        df_pol = metrike_po_grupi(y_test, pred, np.array([gender_names[g] for g in gender_test]), gender_names, "gender")
        df_intersekcija = metrike_po_grupi(y_test, pred, intersekcija_test, intersekcija_imena, "race_x_gender")

        for df in [df_rasa, df_pol, df_intersekcija]:
            df["model"] = ime_modela
            df["klasifikator"] = klasifikator
            sve_fairness_metrike.append(df)

fairness_tabela = pd.concat(sve_fairness_metrike, ignore_index=True)
fairness_tabela.to_csv("/content/drive/MyDrive/diplomski2/results/fairness_tabela.csv", index=False)

def izracunaj_gap(fairness_tabela):
    gapovi = []
    for (model, klasifikator, dimenzija), grupa_df in fairness_tabela.groupby(["model", "klasifikator", "dimenzija"]):
        gap_accuracy = grupa_df["accuracy"].max() - grupa_df["accuracy"].min()
        gap_f1 = grupa_df["f1_macro"].max() - grupa_df["f1_macro"].min()
        gapovi.append({
            "model": model,
            "klasifikator": klasifikator,
            "dimenzija": dimenzija,
            "accuracy_gap": gap_accuracy,
            "f1_gap": gap_f1,
        })
    return pd.DataFrame(gapovi)

gap_tabela = izracunaj_gap(fairness_tabela)
gap_tabela.to_csv("/content/drive/MyDrive/diplomski2/results/fairness_gap_tabela.csv", index=False)
print("\n=== FAIRNESS GAP TABELA ===")
print(gap_tabela.sort_values("accuracy_gap", ascending=False))


# ==============================================================================
# 6. STATISTIČKA ZNAČAJNOST — McNemar Test
# ==============================================================================
def izracunaj_mcnemar(y_true, pred_a, pred_b):
    tocno_a = (pred_a == y_true)
    tocno_b = (pred_b == y_true)
    tabela = pd.crosstab(tocno_a, tocno_b)
    tabela = tabela.reindex(index=[False, True], columns=[False, True], fill_value=0)
    rezultat = mcnemar(tabela.values, exact=False, correction=True)
    return rezultat.pvalue

y_test = np.load("/content/drive/MyDrive/diplomski2/features/convnext_test_age.npy")
pred_convnext = np.load("/content/drive/MyDrive/diplomski2/features/convnext_pred_LogisticRegression.npy")
pred_swin = np.load("/content/drive/MyDrive/diplomski2/features/swin_pred_LogisticRegression.npy")
pred_deit = np.load("/content/drive/MyDrive/diplomski2/features/deit_pred_LogisticRegression.npy")

p_conv_swin = izracunaj_mcnemar(y_test, pred_convnext, pred_swin)
p_conv_deit = izracunaj_mcnemar(y_test, pred_convnext, pred_deit)

print("\n=== STATISTIČKA ZNAČAJNOST (McNemar Test - Logistic Regression) ===")
print(f"ConvNeXt vs Swin: p = {p_conv_swin:.5f} -> {'Statistički značajna razlika' if p_conv_swin < 0.05 else 'Nema statistički značajne razlike'}")
print(f"ConvNeXt vs DeiT: p = {p_conv_deit:.5f} -> {'Statistički značajna razlika' if p_conv_deit < 0.05 else 'Nema statistički značajne razlike'}")


# ==============================================================================
# 7. VIZUELIZACIJA I GRAFICI
# ==============================================================================
# 7.1 Line Plot — Accuracy po rasi
df_rasa_logreg = fairness_tabela[
    (fairness_tabela["dimenzija"] == "race") &
    (fairness_tabela["klasifikator"] == "LogisticRegression")
]

fig, ax = plt.subplots(figsize=(10, 5))
for model in df_rasa_logreg["model"].unique():
    subset = df_rasa_logreg[df_rasa_logreg["model"] == model]
    ax.plot(subset["grupa"], subset["accuracy"], marker="o", label=model)

ax.set_xlabel("Rasna grupa")
ax.set_ylabel("Accuracy")
ax.set_title("Accuracy po rasnoj grupi — Poređenje modela (Logistic Regression)")
ax.legend()
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("/content/drive/MyDrive/diplomski2/results/fairness_race_comparison.png", dpi=300)
plt.show()

# 7.2 Bar Chart — Accuracy po polu
df_pol_logreg = fairness_tabela[
    (fairness_tabela["dimenzija"] == "gender") &
    (fairness_tabela["klasifikator"] == "LogisticRegression")
]

fig, ax = plt.subplots(figsize=(6, 4))
df_pol_logreg.pivot(index="grupa", columns="model", values="accuracy").plot(kind="bar", ax=ax)
ax.set_ylabel("Accuracy")
ax.set_title("Accuracy po polu — Poređenje modela")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("/content/drive/MyDrive/diplomski2/results/fairness_gender_comparison.png", dpi=300)
plt.show()

# 7.3 Confusion Matrica — Dinamičko određivanje najlošije grupe za Swin model
df_swin_rasa = df_rasa_logreg[df_rasa_logreg["model"] == "swin"]
najlosija_rasa_ime = df_swin_rasa.sort_values("accuracy").iloc[0]["grupa"]
najlosija_rasa_idx = race_names.index(najlosija_rasa_ime)

y_test_swin = np.load("/content/drive/MyDrive/diplomski2/features/swin_test_age.npy")
pred_swin_logreg = np.load("/content/drive/MyDrive/diplomski2/features/swin_pred_LogisticRegression.npy")

maska_najlosija = race_test == najlosija_rasa_idx

fig, ax = plt.subplots(figsize=(7, 7))
ConfusionMatrixDisplay.from_predictions(y_test_swin[maska_najlosija], pred_swin_logreg[maska_najlosija], ax=ax)
plt.title(f"Confusion matrica — Swin-B (Najlošija grupa: {najlosija_rasa_ime})")
plt.savefig("/content/drive/MyDrive/diplomski2/results/confusion_najlosija_grupa.png", dpi=300, bbox_inches="tight")
plt.show()

# 7.4 Radar Dijagram
kategorije = ["Accuracy", "F1", "1-MAE (norm.)", "1-Fairness Gap (race)"]
modeli_za_prikaz = ["convnext", "deit", "swin"]
boje = {"convnext": "tab:blue", "deit": "tab:orange", "swin": "tab:green"}
MAX_MAE = 75

def izvuci_vrednosti(ime_modela):
    red_logreg = tabela_rezultata[(tabela_rezultata["model"] == ime_modela) & (tabela_rezultata["klasifikator"] == "LogisticRegression")].iloc[0]
    red_linreg = tabela_rezultata[(tabela_rezultata["model"] == ime_modela) & (tabela_rezultata["klasifikator"] == "LinearRegression")].iloc[0]

    accuracy = red_logreg["accuracy"]
    f1 = red_logreg["f1_macro"]
    mae = red_linreg["mae"]

    gap_red = gap_tabela[
        (gap_tabela["model"] == ime_modela) &
        (gap_tabela["klasifikator"] == "LogisticRegression") &
        (gap_tabela["dimenzija"] == "race")
    ].iloc[0]

    fairness_gap = gap_red["accuracy_gap"]
    return [accuracy, f1, 1 - (mae / MAX_MAE), 1 - fairness_gap]

angles = np.linspace(0, 2 * np.pi, len(kategorije), endpoint=False).tolist()
angles += angles[:1]

fig, ax = plt.subplots(figsize=(7, 7), subplot_kw=dict(polar=True))
for ime_modela in modeli_za_prikaz:
    vrednosti = izvuci_vrednosti(ime_modela)
    vrednosti += vrednosti[:1]
    ax.plot(angles, vrednosti, label=ime_modela, color=boje[ime_modela])
    ax.fill(angles, vrednosti, alpha=0.1, color=boje[ime_modela])

ax.set_xticks(angles[:-1])
ax.set_xticklabels(kategorije)
ax.legend(loc="upper right", bbox_to_anchor=(1.3, 1.1))
plt.title("Poređenje arhitektura — Accuracy, F1, MAE, Fairness")
plt.savefig("/content/drive/MyDrive/diplomski2/results/radar_poredjenje.png", dpi=300, bbox_inches="tight")
plt.show()

print("\nCelokupan eksperiment je uspešno izvršen i svi grafikoni su sačuvani u Google Drive!")
```
